```python
#create a list []
List=[2,20,3,4,1,20,3,5,89,2]
print("Our List is : ",List)
print("Datatype is: ",type(List))

List.append(100)
print("After append 100: ",List)

List.insert(2,50)
print("After insert 50 : ",List)

List.remove(20)
print("After removing 20s first occurence : ",List)

List.count(2)

List.reverse()
print("After reverse :",List)

List.sort()
print("After sorting: ",List)


print("Maximum in List is :",max(List))

print("Minimum in List is :",min(List))

addi=sum(List)
print("Sum of List is :",addi)
```

```python
#creating a tuple()
tup=(1,2,3,12.69)
lis=list(tup)
print("OG tup:",tup)
print("datatype of tup:",type(tup))
print("OG lis:",lis)
print("datatype of lis:",type(lis))
lis.append(69)
print("Updates lis:",lis)
tup=tuple(lis)
print("Updates tup:",tup)

print("Maximum",max(tup))
print(len(tup))
print(tup.count(1))
print(tup.index(3))
```
